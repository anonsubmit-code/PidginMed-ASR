```python
import re
import pandas as pd

def clean_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    # remove percentage signs explicitly
    text = text.replace("%", " ")

    # remove punctuation (.,!?;: etc.)
    text = re.sub(r"[^\w\s]", " ", text)

    # remove extra whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
df["pidgin_transcription"] = df["pidgin_transcription"].apply(clean_text)
df["english_translation"] = df["english_translation"].apply(clean_text)
```

#Code-Switching Annotations

```python
import re
import pandas as pd
from collections import Counter

# Column name
pidgin_col = "pidgin_transcription"

def tokenize(text):
    if pd.isna(text):
        return []

    text = str(text).lower()
    text = re.sub(r"[^a-zA-Z\s]", " ", text)  # remove punctuation/numbers
    text = re.sub(r"\s+", " ", text).strip()

    return text.split()

# tokenize all pidgin transcriptions
df["pidgin_tokens"] = df[pidgin_col].apply(tokenize)

# flatten tokens
all_tokens = [token for tokens in df["pidgin_tokens"] for token in tokens]

# count frequency
token_counts = Counter(all_tokens)

# create unique token dataset
token_df = pd.DataFrame(
    token_counts.items(),
    columns=["token", "frequency"]
)

# sort by frequency
token_df = token_df.sort_values("frequency", ascending=False).reset_index(drop=True)

# add annotation columns
token_df["language_label"] = ""   # e.g., Pidgin / English / Ambiguous / Other
token_df["notes"] = ""

token_df.head()
```
